```python
import pandas as pd
import seaborn as sns

tips = sns.load_dataset("tips")
```

```python
# 문자열로 변환 astype
tips['sex_str'] = tips['sex'].astype(str)
```

```python
print(tips.dtypes)
```

```python
tips['total_bill'] = tips['total_bill'].astype(str)
print(tips.dtypes)
```

```python
tips['total_bill'] = tips['total_bill'].astype(float) #다시 실수로
print(tips.dtypes)
```

```python
# 잘못 입력한 데이터 처리하기
```

```python
tips_sub_miss = tips.head(10)
tips_sub_miss.loc[[1,3,5,7], 'total_bill'] = 'missing'
print(tips_sub_miss)
```

```python
print(tips_sub_miss.dtypes) #missing 떄문에 total_bill 문자열됨
```

```python
tips_sub_miss['total_bill'].astyoe(float) #오류
```

```python
pd.to_numeric(tips_sub_miss['total_bill']) #오류
```

```python
tips_sub_miss['total_bill'] = pd.to_numeric(tips_sub_miss['total_bill'],errors='coerce')
print(tips_sub_miss.dtypes)   #coerce 숫자로 반환할 수 없는 값을 누릭깂으로 지정
```

```python
#다운캐스트(저장하는 실수의 예상 범위가 작다면)
tips_sub_miss['total_bill'] = pd.to_numeric(tips_sub_miss['total_bill'],errors='coerce',downcast='float')
print(tips_sub_miss.dtypes)  #float 64 에서 float 32로 바뀜
```
