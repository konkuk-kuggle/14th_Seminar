```python
def avg_3_apply(col):
    x = col[0]
    y = col[1]
    z = col[2]
    return (x+y+z) /3
print(df.apply(avg_3_apply))
# dafaframe의 행 개수가 3임을 전제
```

```python
def avg_3_apply(col):
    sum = 0
    for item in col: #한 행에 들어있는 모든 열의 값들을 하나씩 꺼내와 sum에 더함
        sum += item
    return sum / df.shape[0]
```

```python
print(df.apply( avg_3_apply))
```

```python
def avg_2_apply(row):
    sum = 0
    for item in row:
        sum += item
    return sum / df.shape[1] #계산한 총합을 전체 열의 개수로 나누어 평균값
print(df.apply(avg_2_apply, axis =1)) #shape[1] 행 방향 데이터 처리
```

```python
#10 -3 apply 메서드 사용하기 - 고급
```

```python
import seaborn as sns
titanic = sns.load_dataset("titanic")
```

```python
print(titanic.head())
```

```python
print(titanic.info())
```

```python
import numpy as np

def count_missing(vec):
    null_vec = pd.isnull(vec) # 누락값 있으면 True(1) 없으면 False(0) 
    null_count = np.sum(null_vec) # 1 개수 더하면 누락값 개수
    return null_count
```

```python
cmis_col = titanic.apply(count_missing)
print(cmis_col)
```

```python
import pandas as pd

def count_missing2(vec):
    # np.sum을 쓰지 않고 Pandas의 .sum()으로 한 번에 처리
    return pd.isnull(vec).sum() 
```

```python
cmis_col = titanic.apply(count_missing2)
print(cmis_col)
```

```python
def count_missing3(vec):
    num = count_missing(vec)
    return num
print(count_missing3(titanic))
```

```python
## 누락값 비율 계산
def prop_missing(vec):
    num = count_missing(vec)
    dem = vec.size
    return num / dem
```

```python

pmis_col = titanic.apply(prop_missing)
print(pmis_col)
```

```python
# 누락값 아닌 데이터 비율 계산
def prop_complete(vec):
    return 1- prop_missing(vec)
```

```python
pmis_col = titanic.apply(prop_complete)
print(pmis_col)
```

```python
# 데이터프레임 누락값 처리 (행 방향 axis=1)
```

```python
cmis_row = titanic.apply(count_missing,axis = 1) #누락값 개수
pmis_row = titanic.apply(prop_missing,axis = 1) #누락값 데이터 비율
pcom_row = titanic.apply(prop_complete,axis = 1) #누락값 아닌 데이터 비율
```

```python
print(cmis_row.head())
```

```python
print(pmis_row.head())
```

```python
print(pcom_row.head())
```

```python
# 누락값 개수 titanic dataframe에 추가
```

```python
titanic['num_missing'] = titanic.apply(count_missing, axis=1)
print(titanic.head())
```

```python
# 누락값이 2이상인 데이터 추출
print(titanic.loc[titanic.num_missing > 1,:].sample(10))
```
