```python
#11 -5. 정규식으로 문자열 처리에 날개 달기
```

```python
import re
tele_num = '1234567890'
```

```python
m = re.match(pattern = '\d\d\d\d\d\d\d\d\d\d', string = tele_num)
print(type(m))  #\d 숫자 1개 #match 패턴 찾아 양 끝 인덱스 반환
```

```python
print(m)
```

```python
if m:
    print('match')
else:
    print('no match')
print(bool(m))
```

```python
print(m.start()) #첫번째 인덱스
```

```python
print(m.end()) #마지막 인덱스
```

```python
print(m.span()) #첫번째 마지막 인덱스
```

```python
print(m.group())
```

```python
tele_num_spaces = '123 456 7890'
m = re.match(pattern = '\d{10}', string=tele_num_spaces)
print(m) #띄어쓰기 있어서 패턴 찾지 못함
```

```python
if m:
    print('match')
else:
    print('no match')
```

```python
p = '\d{3}\s?\d{3}\s?\d{4}'
m = re.match(pattern=p, string=tele_num_spaces)
print(m)
```

```python
tele_num_space_paren_dash = '(123) 456-7890'
p = '\(?\d{3}\)?\s?\d{3}\s?-?\d{4}'
m = re.match(pattern=p, string=tele_num_space_paren_dash)
print(m)
```

```python
cnty_tele_num_space_paren_dash = '+1 (123) 456-7890'
p = '\+?1\s?\(?\d{3}\)?\s?\d{3}\s?-?\d{4}'
m = re.match(pattern=p, string=cnty_tele_num_space_paren_dash)
print(m)
```

```python
#compile() 메서드
#패턴 반복 사용
```

```python
p = re.compile('\d{10}')
s = '1234567890'
m = p.match(s)
print(m)
```
