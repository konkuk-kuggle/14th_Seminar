```python
#3-4 시리즈와 데이터프레임 데이터 변환하기
```

```python
print(scientists.dtypes)
```

```python
born_datetime = pd.to_datetime(scientists['Born'],format = '%Y-%m-%d')
print(born_datetime) #Born 열을 문자열에서 datetime 형으로 변환
```

```python
#새로운 열로 born_datetime, died_datetime을 scientists 데이터프레임에 추가
died_datetime = pd.to_datetime(scientists['Died'], format='%Y-%m-%d')

scientists['born at'],scientists['died_dt'] = (born_datetime, died_datetime)
print(scientists.head())
```

```python
print(scientists.shape)
print(scientists.dtypes)
```

```python
print(scientists['Age'])
```

```python
print(scientists["Age"].sample(frac=1, random_state=42)) #인덱스 번호랑 값이랑 같이 묶어서 랜덤으로
```

```python
scientists["Age"] = scientists["Age"].sample(frac=1, random_state=42)
print(scientists['Age']) #판다스가 대부분의 연산에서 index 자동 병합
```

```python
scientists["Age"] = scientists["Age"].sample(frac=1, random_state=42).values
print(scientists['Age']) #values가 인덱스 번호랑 값이랑 같이 묶어서 랜덤으로 배열된 상태에서 인덱스 없애고 다시 인덱스 번호 매김
```

```python
scientists['age_days'] = (scientists['died_dt'] - scientists['born at'])
print(scientists)
print(type(scientists['age_days']))
```

```python
import numpy as np
scientists['age_years'] = (scientists['age_days'].dt.days / 365).apply(np.floor)  # 날짜 수를 나이로 변환
print(scientists)
# 기간 → 일수 숫자로 변환 → 년으로 변환
# 각각의 값에 floor를 적용 → 소수점 버림
```

```python
# age_days의 값들은 숫자가 아니라 Timedata 자료형
# 따라서 .dt는 날짜/시간 관련 데이터에서 그 안의 요소(숫자)를 꺼내는 접근자
```

```python
scientists = scientists.assign(
    age_days_assign=scientists['died_dt'] - scientists['born at'],
    age_year_assign=(scientists['age_days'].dt.days / 365).apply(np.floor)
)
print(scientists) #assign() 메서드 열 할당하고 수정
```

```python
scientists = scientists.assign(
    age_days_assign=scientists["died_dt"] - scientists["born at"],
    age_year_assign=lambda df_: (df_["age_days_assign"].dt.days / 365).apply(np.floor), 
)
print(scientists)
#age_days가 없을 떄 lambda() 이용해서
```
