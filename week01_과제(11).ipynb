```python
#agg 메서드로 사용자 함수와 groupby 메서드 조합하기
```

```python
def my_mean(values):
    n = len(values)
    sum=0
    for value in values:
        sum += value
    return sum / n
```

```python
agg_my_mean = df.groupby('year')['lifeExp'].agg(my_mean)
print(agg_my_mean)
```

```python
#2개의 인잣값을 받아 처리하는 사용자 함수와 groupby 메서드
```

```python
def my_mean_diff(values, diff_value):
    n = len(values)
    sum = 0
    for value in values:
        sum += value
    mean = sum / n #mean 연도별 평균 수명 구함
    return mean - diff_value
```

```python
# 전체 평균 수명
global_mean = df.lifeExp.mean()
print(global_mean)
```

```python
agg_mean_diff = df.groupby('year').lifeExp.agg(my_mean_diff, diff_value=global_mean)
print(agg_mean_diff)
```

```python
import seaborn as sns
import numpy as np
```

```python
np.random.seed(42) #랜덤 결과를 매번 똑같이 나오게 고정
```

```python
# 임의로 4개 선택해 누락값으로 바꿈
tips_10 = sns.load_dataset('tips').sample(10)
tips_10.loc[np.random.permutation(tips_10.index)[:4], 'total_bill'] = np.nan
print(tips_10)      #index 순서 랜덤하게 섞음     #섞은 것 중 4개 가져옴
```

```python
# 여성과 남성 구별해 total_bill 열 평균값 구하기
count_sex = tips_10.groupby('sex').count()
print(count_sex) #남성 누락값 4개 여성 누락값 없음
```

```python
# 성별 구분해 total_bill 열 데이터 받아 평균값 구하는 함수
def fill_na_mean(x):
    avg = x.mean()
    return x.fillna(avg)
```

```python
total_bill_group_mean = tips_10.groupby('sex').total_bill.transform(fill_na_mean)
tips_10['fill_total_bill'] = total_bill_group_mean
print(tips_10)
```

```python
#그룹 오브젝트 살펴보기
```

```python
tips_10 = sns.load_dataset('tips').sample(10,random_state = 42)
print(tips)
```

```python
grouped = tips_10.groupby('sex')
print(grouped)
```

```python
# 그룹 오브젝트에 포함된 속성보기
print(grouped.groups)
```

```python
#그룹 오브젝트의 평균 구하기
```

```python
avgs = grouped.mean(numeric_only=True)
print(avgs)
```

```python
print(tips_10.columns) # 평균 값 구할 수 없는 열인 smoker,day,time 등등 그룹 연산에서 자동 제외됨
```

```python
#그룹 오브젝트에서 데이터 추출하고 반복하기
```

```python
# 그룹 오브젝트에서 성별 여성만 추출함 get_group()
female = grouped.get_group('Female')
print(female)
```

```python
for sex_group in grouped:
    print(sex_group)
```

```python
for sex_group in grouped:
    # 객체 자료형 (튜플)
    print(f'the type is: {type(sex_group)}\n')

    # 객체 길이 (2)
    print(f'the length is: {len(sex_group)}\n')

    # 첫 번째 요소
    first_element = sex_group[0]
    print(f'the first element is: {first_element}\n')

    # 첫 번째 요소의 자료형(문자열)
    print(f'it has a type of: {type(sex_group)}\n')

    # 두 번째 요소
    second_element = sex_group[1]
    print(f'the second element is:\n{second_element}\n')

    # 두 번째 요소의 자료형 (데이터프레임)
    print(f'it has a type of: {type(second_element)}\n')

    # 그룹 출력
    print(f'what we have:')
    print(sex_group)

    # for문 중단
    break
```
