# 1. 데이터 로드

```python
from sklearn.datasets import load_breast_cancer

data = load_breast_cancer()
x, y = data.data, data.target

print(f"전체 데이터 형태 - 특징(x): {x.shape}, 정답(y): {y.shape}")
```

# 2. 데이터 분할

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)

print(f"학습용 데이터 개수: {x_train.shape[0]}개")
print(f"테스트용 데이터 개수: {x_test.shape[0]}개")
```

# 3. 스케일링

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

# 4. 모델 생성

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
```

Q1. 빈칸 채우기 (모델 객체 생성 함수)

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier

models = {
    "로지스틱 회귀": LogisticRegression(random_state=42),
    "KNN": KNeighborsClassifier(),
    "의사결정 나무": DecisionTreeClassifier(random_state=42),
    "랜덤 포레스트": RandomForestClassifier(random_state=42),
}
```

# 5. 모델 학습 및 예측

Q2. 빈칸 채우기 (모델 학습. 예측)

```python
predictions = {}

for name, model in models.items():
  model.fit(x_train_scaled, y_train)
  pred = model.predict(x_test_scaled)
  predictions[name] = pred
  print(f"[{name}] 학습 및 예측 완료")
```

# 6. 모델 평가

```python
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    f1_score,
    precision_score,
    recall_score,
)

for name, model in models.items():
  y_pred = predictions[name]

  cm = confusion_matrix(y_test, y_pred)
  acc = accuracy_score(y_test, y_pred)
  prec = precision_score(y_test, y_pred)
  rec = recall_score(y_test, y_pred)
  f1 = f1_score(y_test, y_pred)

  print(f"================ {name} 평가 결과 ================")
  print(f"[혼동 행렬 (Confusion Matrix)]\n{cm}\n")
  print(f"• 정확도 (Accuracy)  : {acc:.4f}")
  print(f"• 정밀도 (Precision) : {prec:.4f}")
  print(f"• 재현율 (Recall)    : {rec:.4f}")
  print(f"• F1-score         : {f1:.4f}\n")
```
