# join, splitlines, replace 메서드 실습하기

```python
d1='40'
m1="46'"
s1='52.837"'
u1='N'

d2 = '73' 
m2 = "58'" 
s2 = '26.302"' 
u2 = 'W'

coords=' '.join([d1,m1,s1,u1,d2,m2,s2,u2])
print(coords)
```

```python
multi_str="""Guard: What? Ridden on a horse?
king Arthur: Yes!
"""
print(multi_str)
```

```python
multi_str_split=multi_str.splitlines()
print(multi_str_split)
```

```python
print(multi_str_split[::2])
```

```python
print(multi_str.replace("Guard: ","").splitlines()[::2])
```

# 문자열 포매팅 실습하기

```python
var = 'flesh wound' 
s = "It's just a {}!"

print(s.format(var))
print(s.format('scratch'))
```

# 수치값 포매팅 실습

```python
print('Some digits of pi: {}'.format(3.14159265359))
```

# % 연산자로 포매팅하기

```python
s='I only know %d digits of pi'% 7
print(s)
```

```python
print('some digits of %(cont)s: %(value).2f' % {'cont':'e','value':2.718})
```

# f-strings로 포매팅 사용하기

```python
var = 'flesh wound'
s=f"It's just a {var}!"
print(s)
```

# 정규식으로 전화번호 패턴 찾기

```python
import re
tele_num='1234567890'
m=re.match(pattern='\d\d\d\d\d\d\d\d\d',string=tele_num)
print(type(m))
```
